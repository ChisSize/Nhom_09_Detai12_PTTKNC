# Chạy toàn bộ pipeline — Đề tài 12
Notebook này chạy lần lượt 5 notebook trong `src/` theo đúng thứ tự, ghi lại output và hình vẽ. Chạy từ thư mục gốc dự án hoặc từ `notebooks/`.

```python
from pathlib import Path
import nbformat
from nbconvert.preprocessors import ExecutePreprocessor

SRC = (Path.cwd() / "src") if (Path.cwd() / "src").exists() else (Path.cwd().parent / "src")
ORDER = [
    "01_data_cleaning.ipynb",
    "02_assumption_checks.ipynb",
    "03_manova.ipynb",
    "04_visualization.ipynb",
    "05_confidence_ellipses.ipynb",
    "06_sensitivity.ipynb",
]

for nb_name in ORDER:
    nb_path = SRC / nb_name
    print(f"=== Bắt đầu: {nb_name} ===")

    notebook = nbformat.read(str(nb_path), as_version=4)
    executor = ExecutePreprocessor(timeout=900)
    executor.preprocess(notebook, {"metadata": {"path": str(SRC)}})
    nbformat.write(notebook, str(nb_path))

    print(f"=== Hoàn thành: {nb_name} ===\n")

print("HOÀN THÀNH TOÀN BỘ PIPELINE")
```
