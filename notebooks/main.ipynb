# Chạy toàn bộ pipeline — Đề tài 12
Notebook này chạy lần lượt 5 notebook trong `src/` theo đúng thứ tự, ghi lại output và hình vẽ. Chạy từ thư mục gốc dự án hoặc từ `notebooks/`.

```python
import subprocess, sys
from pathlib import Path

SRC = (Path.cwd() / "src") if (Path.cwd() / "src").exists() else (Path.cwd().parent / "src")
ORDER = ["01_data_cleaning.ipynb", "02_assumption_checks.ipynb", "03_manova.ipynb", "04_visualization.ipynb", "05_confidence_ellipses.ipynb"]

for nb_name in ORDER:
    print(f"=== Bắt đầu: {nb_name} ===")
    subprocess.run([sys.executable, "-m", "jupyter", "nbconvert", "--to", "notebook", "--execute", "--inplace", nb_name], cwd=SRC, check=True)
    print(f"=== Hoàn thành: {nb_name} ===\n")

print("HOÀN THÀNH TOÀN BỘ PIPELINE")
```
